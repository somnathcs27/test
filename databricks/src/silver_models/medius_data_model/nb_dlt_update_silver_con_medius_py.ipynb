```python
import dlt
from mdp_databricks_common.sql.sql_loader import SQLLoader
from mdp_databricks_common.utils.data_load_utils import get_asset_bundle_root_path

env_var = dbutils.secrets.get(scope = 'data-kv-scope', key = 'data-platform-environment')
job_name = "MEDIUS"
```

#Demand Capture

```python
expectations = {
    "row_hash_is_not_null": "MDP_ROW_HASH IS NOT NULL"
}

@dlt.table(
    name="medius_demand_capture",
    comment="Silver Layer table for Medius Demand Capture."
)
@dlt.expect_all_or_drop(expectations)
def silver_con_spectra_users():
    sl = SQLLoader(
        environment_name=env_var,
        target_path=(
            f"{get_asset_bundle_root_path(env_var)}"
            "files/src/silver_models/medius_data_model"
            "/sql_scripts/silver/confidential/medius_demand_capture.sql"
        ),
        replacements=None
    )
    return spark.sql(
        sl.load()
    )
```
