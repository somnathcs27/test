```python
import dlt
from mdp_databricks_common.sql.sql_loader import SQLLoader
from mdp_databricks_common.utils.data_load_utils import get_asset_bundle_root_path

env_var = dbutils.secrets.get(scope = 'data-kv-scope', key = 'data-platform-environment')
job_name = "SPECTRA"
```

#Spectra Companies

```python
expectations = {
    "primary_keys_are_not_null": "COMPANIES_ID IS NOT NULL",
    "row_hash_is_not_null": "MDP_ROW_HASH IS NOT NULL"
}

@dlt.table(
    name="spectra_companies",
    comment="Silver Layer table for Spectra Companies."
)
@dlt.expect_all_or_drop(expectations)
def silver_int_spectra_companies():
    sl = SQLLoader(
        environment_name=env_var,
        target_path=(
            f"{get_asset_bundle_root_path(env_var)}"
            "files/src/silver_models/spectra_data_model"
            "/sql_scripts/silver/internal/spectra_companies.sql"
        ),
        replacements=None
    )
    return spark.sql(
        sl.load()
    )
```
